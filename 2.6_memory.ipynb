```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
import os

llm = init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=os.getenv("AGENT_API_KEY"),
    base_url=os.getenv("BASE_URL"),
)

agent = create_agent(
    model=llm
)
```

加入记忆功能

```python
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
import os

llm = init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=os.getenv("AGENT_API_KEY"),
    base_url=os.getenv("BASE_URL"),
)

agent = create_agent(
    model=llm,
    checkpointer=InMemorySaver()
)
```

```python
from langchain_core.messages import HumanMessage

config = {"configurable": {"thread_id": "thread_1"}}

response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="你好，我叫xxx，我喜欢猫猫")
        ]
    },
    config
)

for chunk in response["messages"]:
    chunk.pretty_print()
```

```python
response = agent.invoke(
    {
        "messages": (
            HumanMessage(content="你知道我喜欢什么动物吗？")
        )
    },
    config
)

for chunk in response["messages"]:
    chunk.pretty_print()
```

Memory持久化存储
    导入依赖
    初始化checkpointer
    自动建表
    创建Agent，指定checkpointer

```python
import sqlite3
from langgraph.checkpoint.sqlite import SqliteSaver

from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
import os

connection = sqlite3.connect("resources/checkpoint.db", check_same_thread=False)
checkpointer = SqliteSaver(connection)

checkpointer.setup()
llm = init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=os.getenv("AGENT_API_KEY"),
    base_url=os.getenv("BASE_URL"),
)

agent = create_agent(
    model=llm,
    checkpointer=checkpointer
)
```

```python
from langchain_core.messages import HumanMessage

config = {
    "configurable": {
        "thread_id": "thread_3"
    }
}

response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="你好我叫xxx，我喜欢狗")
        ]
    },
    config
)

for chunk in response["messages"]:
    chunk.pretty_print()
```

```python
from langchain_core.messages import HumanMessage


config = {
    "configurable": {
        "thread_id": "thread_1"
    }
}

response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="你知道我喜欢什么动物吗")
        ]
    },
    config
)

for chunk in response["messages"]:
    chunk.pretty_print()
```
